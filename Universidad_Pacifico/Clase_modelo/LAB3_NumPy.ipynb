## ¿Qué es Numpy?

NumPy es una biblioteca de Python que se utiliza para trabajar con arreglos, álgebra lineal, transformada de Fourier y matrices. NumPy significa Python Numérico y es un proyecto de código abierto. El objeto de arreglo en NumPy se llama **ndarray** y proporciona muchas funciones de soporte que facilitan enormemente el trabajo con ndarray.

Los arreglos se utilizan con mucha frecuencia en la ciencia de datos, donde la velocidad y el uso de recursos son cruciales.

NumPy se suele importar con el alias `np`.

Generalmente tiene un tamaño fijo y cada elemento es del mismo tipo. Podemos convertir una lista en un arreglo de NumPy importando primero `numpy`:

<hr>

```python
# import numpy library

import numpy as np 
```

A continuación, convertimos la lista de la siguiente manera:

```python
# Create a numpy array

a = np.array([0, 1, 2, 3, 4])
a
```

Al igual que con las listas, podemos acceder a cada elemento mediante un corchete:

```python
# Imprima cada elemento

print("a[0]:", a[0])
print("a[1]:", a[1])
print("a[2]:", a[2])
print("a[3]:", a[3])
print("a[4]:", a[4])
```

### Slicing (segmentación)

Al igual que con las listas, podemos segmentar un array de NumPy. En Python, segmentar significa tomar los elementos desde un índice dado hasta otro índice dado.

Pasamos la segmentación de esta manera: [inicio:fin]. El elemento en el índice final no se incluye en la salida.

Podemos seleccionar los elementos del 1 al 3 y asignarlos a un nuevo array de NumPy `d` de la siguiente manera:

```python
# Segmentación del array numpy

d = c[1:4]
d
```

Podemos asignar los índices correspondientes a los nuevos valores de la siguiente manera:

```python
# Set the fourth element and fifth element to 300 and 400

c[3:5] = 300, 400
c
```

También podemos definir los pasos en el seccionamiento, de esta manera: [inicio:fin:paso].

```python
arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[1:5:2])
```

Si no pasamos el inicio, se considera 0.

```python
print(arr[:4])
```

Si no pasamos el final, considera hasta la longitud del array.

```python
print(arr[4:])
```

Si no pasamos el paso se considera 1

```python
print(arr[1:5:])
```

### Adición de matrices

Consider the numpy array <code>u</code>:

```python
u = np.array([1, 0])
u
```

Consideremos el array de numpy <code>v</code>:

```python
v = np.array([0, 1])
v
```

Podemos sumar los dos arreglos y asignarlos a z:

```python
# Numpy Array Addition

z = np.add(u, v)
z
```

The operation is equivalent to vector addition:

```python
# Plotting functions


import time 
import sys
import numpy as np 

import matplotlib.pyplot as plt


def Plotvec1(u, z, v):
    
    ax = plt.axes() # to generate the full window axes
    ax.arrow(0, 0, *u, head_width=0.05, color='r', head_length=0.1)# Add an arrow to the  U Axes with arrow head width 0.05, color red and arrow head length 0.1
    plt.text(*(u + 0.1), 'u')#Adds the text u to the Axes 
    
    ax.arrow(0, 0, *v, head_width=0.05, color='b', head_length=0.1)# Add an arrow to the  v Axes with arrow head width 0.05, color red and arrow head length 0.1
    plt.text(*(v + 0.1), 'v')#Adds the text v to the Axes 
    
    ax.arrow(0, 0, *z, head_width=0.05, head_length=0.1)
    plt.text(*(z + 0.1), 'z')#Adds the text z to the Axes 
    plt.ylim(-2, 2)#set the ylim to bottom(-2), top(2)
    plt.xlim(-2, 2)#set the xlim to left(-2), right(2)
```

```python
# Plot numpy arrays

Plotvec1(u, z, v)
```

## Linspace

Una función útil para graficar funciones matemáticas es `linspace`. `linspace` devuelve números espaciados uniformemente en un intervalo especificado.

**numpy.linspace(start, stop, num = valor entero)**

start: inicio del intervalo

stop: fin del intervalo

num: número de muestras a generar.

```python
# Makeup a numpy array within [-2, 2] and 5 elements

np.linspace(-2, 2, num=5)
```

Si cambiamos el parámetro <code>num</code> a 9, obtenemos 9 números espaciados uniformemente en el intervalo de -2 a 2:

```python
# Make a numpy array within [-2, 2] and 9 elements

np.linspace(-2, 2, num=9)
```

Podemos usar la función <code>linspace</code> para generar 100 muestras igualmente espaciadas del intervalo de 0 a 2π:

```python
# Make a numpy array within [0, 2π] and 100 elements 

x = np.linspace(0, 2*np.pi, num=100)
```

Podemos aplicar la función seno a cada elemento del array <code>x</code> y asignarlo al array <code>y</code>:

```python
# Calculate the sine of x list

y = np.sin(x)
```

```python
# Plot the result

plt.plot(x, y,color='k')
```

### Iteración de arreglos unidimensionales

Iterar significa recorrer los elementos uno por uno.

Si iteramos sobre un arreglo unidimensional, recorreremos cada elemento individualmente.

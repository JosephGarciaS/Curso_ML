## Visualización y acceso a los datos

También puedes obtener una columna como una serie. Puedes pensar en una serie de Pandas como un dataframe unidimensional. Solo usa un corchete:

```python
# Obtén la columna como una serie

x = df['Producto']
x
```

También puedes obtener una columna como un dataframe. Por ejemplo, podemos asignar la columna **Cantidad**:

```python
# Obtener la columna como un DataFrame
x = df[['Cantidad']]
type(x)
```

Puedes hacer lo mismo para varias columnas; simplemente colocamos el nombre del dataframe, en este caso, <code>df</code>, y el nombre de los encabezados de las columnas entre corchetes dobles. El resultado es un nuevo dataframe compuesto por las columnas especificadas:

```python
# Access to multiple columns

y = df[['Producto','Categoria', 'Cantidad']]
y
```

Una forma de acceder a elementos únicos es mediante el método <code>iloc</code>, donde se puede acceder a la primera fila y a la primera columna de la siguiente manera:

```python
# Acceda al valor de la primera fila y la primera columna.

df.iloc[0, 0]
```

```python
# Acceda al valor de la segunda fila y la primera columna.

df.iloc[1,0]
```

```python
# Acceda al valor de la primera fila y la tercera columna.
df.iloc[0,2]
```

```python
# Acceda al valor de la segunda fila y la tercera columna.
df.iloc[2,2] # corregir
```

También puede acceder a la columna utilizando el nombre; los siguientes son los mismos que los anteriores:

```python
# Access the column using the name

df.loc[0, 'PedidoID']
```

```python
# Access the column using the name

df.loc[1, 'PedidoID']
```

```python
# Access the column using the name

df.loc[0, 'Categoria']
```

```python
# Access the column using the name

df.loc[1, 'Categoria']
```

Puede realizar la segmentación utilizando tanto el índice como el nombre de la columna:

```python
# Segmenta el dataframe

df.iloc[0:2, 0:3]
```

```python
# Segmentando el dataframe utilizando los nombres

df.loc[0:2, 'PedidoID':'Categoria']
```

## Exercise 1: Pandas: DataFrame and Series
**Pandas** es una popular biblioteca para el análisis de datos, basada en el lenguaje de programación Python. Pandas generalmente proporciona dos estructuras de datos para manipular datos:
* DataFrame
* Series
  
Un **DataFrame** es una estructura de datos bidimensional, es decir, los datos se organizan en filas y columnas de forma tabular.

* Un DataFrame de Pandas se crea cargando conjuntos de datos desde un almacenamiento existente.
* El almacenamiento puede ser una base de datos SQL, un archivo CSV, un archivo de Excel, etc.
* También se puede crear a partir de listas, diccionarios o listas de diccionarios.

Una **Serie** representa una matriz unidimensional de datos indexados. Tiene dos componentes principales:

1. Una matriz de datos propiamente dicha.
2. Una matriz asociada de índices o etiquetas de datos.

El índice se utiliza para acceder a valores de datos individuales. También se puede obtener una columna de un DataFrame como una **Serie**. Puedes pensar en una serie de Pandas como un dataframe unidimensional.

### Ejercicio 3: Segmentación (Slicing)
La segmentación utiliza el operador [] para seleccionar un conjunto de filas y/o columnas de un DataFrame.

Para segmentar un conjunto de filas, se utiliza la siguiente sintaxis: data[inicio:fin],

donde inicio representa el índice desde el que se considera, y fin representa el índice inmediatamente posterior a la fila que se desea seleccionar. Se puede realizar la segmentación utilizando tanto el índice como el nombre de la columna.

> NOTA: Al segmentar en pandas, el límite de inicio se incluye en la salida.

Por lo tanto, si se desean seleccionar las filas 0, 1 y 2, el código sería el siguiente: df.iloc[0:3].

Esto significa que se le indica a Python que comience en el índice 0 y seleccione las filas 0, 1, 2 hasta la 3 (sin incluirla).

> NOTA: Las etiquetas deben encontrarse en el DataFrame; de ​​lo contrario, se producirá un error de clave (KeyError).

La indexación por etiquetas (es decir, usando `loc()`) difiere de la indexación por números enteros (es decir, usando `iloc()`). Con `loc()`, tanto el límite inicial como el límite final son inclusivos. Al usar `loc()`, se pueden usar números enteros, pero estos se refieren a la etiqueta del índice y no a la posición.

Por ejemplo, usar `loc()` para seleccionar las filas 1 a 4 dará un resultado diferente que usar `iloc()` para seleccionar las filas 1 a 4.

##### También podemos seleccionar un valor de datos específico usando la ubicación de fila y columna dentro del DataFrame y la indexación `iloc`.

```python
# Vamos a realizar el seccionamiento utilizando el antiguo dataframe df.

df.iloc[0:2, 0:3]
```

```python
# Vamos a realizar el seccionamiento usando la función loc() en el antiguo dataframe df donde la columna de índice tiene las etiquetas 0,1,2.

df.loc[0:2,'ID_Cliente':'Ciudad']
```

```python
# Vamos a realizar el seccionamiento usando la función loc() en el nuevo dataframe df2 donde la columna de índice es Nombre y tiene las etiquetas: Rose, John y Jane.
df2.loc['Carlos Reyes':'Andrea Castro', 'ID_Estudiante':'Edad']
```

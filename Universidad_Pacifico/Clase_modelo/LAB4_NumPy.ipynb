## Crear una matriz Numpy 2D

```python
# Import the libraries

import numpy as np
```

Consideremos la lista <code>a</code>, que contiene tres listas anidadas **cada una de igual tamaño**.

```python
# Create a list

a = [[11, 12, 13], [21, 22, 23], [31, 32, 33]]
a
```

Podemos convertir la lista a un array de Numpy de la siguiente manera:

```python
# Convert list to Numpy Array
# Every element is the same type

A = np.array(a)
A
```

Podemos usar el atributo <code>ndim</code> para obtener el número de ejes o dimensiones, denominado rango.

```python
# Show the numpy array dimensions

A.ndim
```

El atributo <code>shape</code> devuelve una tupla que corresponde al tamaño o número de cada dimensión.

```python
# Show the numpy array shape

A.shape
```

El número total de elementos en el array viene dado por el atributo <code>size</code>.

```python
# Show the numpy array size

A.size
```

## Operaciones básicas

También podemos sumar matrices. El proceso es idéntico a la suma de matrices. La suma de matrices <code>X</code> e <code>Y</code> se muestra en la siguiente figura:

<img src="https://github.com/JosephGarciaS/Curso_ML/blob/main/Universidad_Pacifico/Clase_modelo/Fig_5.png?raw=true" width="300">

El array de NumPy viene dado por <code>X</code> y <code>Y</code>.

```python
# Create a numpy array X

X = np.array([[1, 0], [0, 1]]) 
X
```

```python
# Create a numpy array Y

Y = np.array([[2, 1], [1, 2]]) 
Y
```

Podemos agregar los arrays de numpy de la siguiente manera.

```python
# Add X and Y

Z = X + Y
Z
```

Multiplicar un array de NumPy por un factor de escala es idéntico a multiplicar una matriz por un factor de escala. Si multiplicamos la matriz <code>Y</code> por el factor de escala 2, simplemente multiplicamos cada elemento de la matriz por 2, como se muestra en la figura.

<img src="https://github.com/JosephGarciaS/Curso_ML/blob/main/Universidad_Pacifico/Clase_modelo/Fig_6.png?raw=true" width="300">

Podemos realizar la misma operación en numpy de la siguiente manera:

```python
# Create a numpy array Y

Y = np.array([[2, 1], [1, 2]]) 
Y
```

```python
# Multiply Y with 2

Z = 2 * Y
Z
```

La multiplicación de dos matrices corresponde a un producto elemento a elemento o producto de Hadamard. Consideremos las matrices X e Y. El producto de Hadamard consiste en multiplicar cada uno de los elementos que se encuentran en la misma posición, es decir, multiplicar los elementos contenidos en las mismas casillas de color. El resultado es una nueva matriz del mismo tamaño que la matriz Y o X, como se muestra en la siguiente figura.

<img src="https://github.com/JosephGarciaS/Curso_ML/blob/main/Universidad_Pacifico/Clase_modelo/Fig_7.png?raw=true" width="300">

Podemos realizar el producto elemento a elemento de la matriz <code>X</code> y <code>Y</code> de la siguiente manera:

```python
# Create a numpy array Y

Y = np.array([[2, 1], [1, 2]]) 
Y
```

```python
# Create a numpy array X

X = np.array([[1, 0], [0, 1]]) 
X
```

```python
# Multiply X with Y

Z = X * Y
Z
```

También podemos realizar la multiplicación de matrices con los arreglos numpy <code>A</code> y <code>B</code> de la siguiente manera:

Primero, definimos las matrices <code>A</code> y <code>B</code>:

```python
# Create a matrix A

A = np.array([[0, 1, 1], [1, 0, 1]])
A
```

```python
# Create a matrix B

B = np.array([[1, 1], [1, 1], [-1, 1]])
B
```

Usamos la función numpy <code>dot</code> para multiplicar los arrays entre sí.

```python
# Calculate the dot product

Z = np.dot(A,B)
Z
```

```python
# Calculate the sine of Z

np.sin(Z)
```

Utilizamos el atributo numpy <code>T</code> para calcular la matriz transpuesta.

```python
# Create a matrix C

C = np.array([[1,1],[2,2],[3,3]])
C
```

```python
# Get the transposed of C

C.T
```
